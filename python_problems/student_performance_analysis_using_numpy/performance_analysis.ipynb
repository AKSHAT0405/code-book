### Project Title: "Student Performance Analysis Using Numpy"

```python
import numpy as np
```

##### given student marks in 2d matrix  <br>
##### Rows : students(S1 to S5) , Columns : Subjects (Math , Science ,English , History)

```python
# rows:students , columns : subjects 
marks = np.array([
    (88,92,85,78),
    (76,85,90,89),
    (92,88,95,94),
    (65,70,72,68),
    (90,95,91,99)
])
```

```python
# task 1 : total & average marks per students

total_marks = np.sum(marks,axis=1)
for i , mark in enumerate(total_marks):
    print(f"S_{i+1} total marks: ",mark)

print("\n")

average_marks = np.mean(marks,axis=1)
for j,m in enumerate(average_marks):
    print(f"S_{j+1} average marks: ", m)
```
